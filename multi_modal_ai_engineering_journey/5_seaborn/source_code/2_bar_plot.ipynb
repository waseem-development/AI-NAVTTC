```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset("penguins").sample(5)
```

```python
df.describe()
```

```python
df.columns.tolist()
```

```python
sns.barplot(x=df.island, y=df.bill_length_mm)
plt.show()
```

```python
sns.set_theme(style="darkgrid")
order = ["Biscoe", "Dream", "Torgersen"]

sns.barplot(
    data=df,
    x="island",
    y="bill_length_mm",
    hue="sex",
    palette="rainbow",
    order=order,
    legend=True,
    hue_order=["Female", "Male"],
    errorbar=("ci", 100),  # replaces ci=100 (old name); 100% confidence interval, default is 95
    # n_boot=2,
    orient="v",  # "v" = vertical bars (x is categorical); for horizontal bars use "h" and swap x and y
    # saturation=0.6,
    err_kws={"color": "b", "linewidth": 5},  # replaces errcolor="b" and errwidth=5
    capsize=0.5,
    # dodge=False,
    # alpha=0.5,
)

plt.legend(loc="lower left")

plt.show()
```

```python
sns.barplot(
    data=df,
    x="bill_depth_mm",
    y="bill_length_mm", 
    # color="green",
    palette="Accent",
    orient="h", # horizontal bars are only valid for numerical data on both axis

)

plt.show()
```
