```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=sns.load_dataset("penguins").sample(50)
```

```python
markers_dict = {
    "Male": "^",
    "Female": "o"
}

sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="sex",
    style="sex",               # needed, or markers is ignored
    palette="viridis",
    size="sex",
    sizes=(100, 90),
    # alpha=0.4
    markers=markers_dict
)

plt.legend(loc="lower right")
plt.show()
```
