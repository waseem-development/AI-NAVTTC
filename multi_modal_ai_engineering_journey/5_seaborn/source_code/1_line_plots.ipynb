```python
sns.lineplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    label="Bill depth vs length",
)
plt.show()
```

```python
sns.lineplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="sex",
    # size=100,
    style="sex",
    palette="rocket",
    markers=["o", "^"],
    dashes=[(4, 2), (1, 3)], # dashes can be true, false, 
    markersize=9,
    errorbar=None,   # removes the shaded band so it's less messy
    legend=True # for legend we have "full", "auto", "brief" or a boolean (True / False)
)

plt.grid()
plt.title("Male VS Female Penguin \"Bill Length and Bill Depth")
plt.legend(loc="lower right")
plt.show()
```

#### Dashes Parameters

| Pattern | Looks like |
|---|---|
| `(1, 0)` | solid line (no gap) |
| `(4, 2)` | `— — — —` long dashes |
| `(2, 2)` | `- - - - -` short dashes |
| `(1, 3)` | `· · · ·` dotted look |
