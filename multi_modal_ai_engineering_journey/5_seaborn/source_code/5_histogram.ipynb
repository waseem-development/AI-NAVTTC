```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pds
```

```python
df = sns.load_dataset("penguins")
df
```

```python
sns.histplot(
    data=df,
    x="flipper_length_mm",
    bins=[170, 180, 190, 200, 210, 220, 230, 240],
    kde=True,  # Kernel Density Estimator
    color="orange",
)
sns.rugplot(data=df, x="flipper_length_mm")
plt.show()
```

```python
sns.histplot(
    data=df,
    kde=True,  # Kernel Density Estimator
    color="orange",
    log_scale=True
)
sns.rugplot(data=df, x="flipper_length_mm")
plt.show()
```
