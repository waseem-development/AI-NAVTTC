```python
annotstyle = {
    "fontsize": 30, 
    "color": "white"
}
labels = np.array([
    ["a0", "a1", "a2", "a3", "a4"],
    ["b0", "b1", "b2", "b3", "b4"]
])

values = np.arange(1, 11).reshape(2, 5)   # numbers 1-10, same shape as labels

v = sns.heatmap(
    values,
    vmin=1,
    vmax=10,
    cmap="rocket",
    annot=labels,
    annot_kws=annotstyle,
    fmt="s",
    linewidths=5,
    linecolor="teal",
    cbar=False,
    xticklabels=False,
    yticklabels=False
)

v.set(
    xlabel="python",
    ylabel="codebpaxto"
)

sns.set_theme(font_scale=5)
plt.show()
```
