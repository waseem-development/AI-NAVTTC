# 5 Number Summary

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
ds = pd.Series([2,4,7,10, 15, 20, 25, 205])
ds
```

5 Number summary includes:
- Minimum Value
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum Value

```python
minimum = ds.min()
q1 = ds.quantile(0.25, interpolation='midpoint')
median = ds.median()
q3 = ds.quantile(0.75, interpolation='midpoint')
maximum = ds.max()
dRange = ds.max() - ds.min()

print(f'minimum: {minimum}')
print(f'q1: {q1}')
print(f'median: {median}')
print(f'q3: {q3}')
print(f'maximum: {maximum}')
print(f'Range: {dRange}')
```

```python
iqr = q3 - q1
print(f"\nStep 4 — IQR ({q3} - {q1}): {iqr}")

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers = ds[(ds < lower_bound) | (ds > upper_bound)]

print(f"Outlier Values: {outliers.values if len(outliers) > 0 else 'None'}")
```

```python
sns.boxplot(data=ds)
```

```python
l = []
for i in ds:
    if ((i >= lower_bound) & (i <= upper_bound)):
        l.append(i)
ds = l
ds
```

```python
sns.boxplot(data=ds)
```
