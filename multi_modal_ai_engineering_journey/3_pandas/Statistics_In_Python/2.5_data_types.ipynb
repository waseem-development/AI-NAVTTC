# Data Types in Stats

```python
import pandas as pd

data_nominal = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female", "Male", "Female"],
    "Blood_Type": ["A", "B", "O", "AB", "A", "O", "B", "AB"],
    "Country": ["Pakistan", "India", "Pakistan", "USA", "UK", "India", "USA", "UK"]
})

print(data_nominal)
```

## How many males and females

```python
data_nominal["Gender"].value_counts()
```

## Find most common blood type:

```python
data_nominal["Blood_Type"].value_counts()
```

## Convert to categorical:

🧠 What’s Actually Happening

When you create this:

"Gender": ["Male", "Female", ...]

Pandas stores it as:

dtype = object  (basically string)

👉 That means:

Python sees it as plain text
Not as a “category” with structure
⚠️ Why This Matters

If you leave it as object:

❌ Slower memory usage
❌ No category-specific optimizations
❌ Harder to control categories (like ordering, encoding)
✅ When You Do This:
data_nominal["Gender"] = data_nominal["Gender"].astype("category")

Now:

✔ Stored efficiently
✔ Recognized as categorical data
✔ Ready for ML preprocessing
✔ Easier encoding later (very important)

```python
data_nominal["Gender"] = data_nominal["Gender"].astype("category")
data_nominal.dtypes
```

## 2. Ordinal Data Practice (Has Order)
📊 Example Dataset: Student Feedback

```python
data_ordinal = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Satisfaction": ["Good", "Excellent", "Average", "Good", "Poor", "Excellent", "Average", "Good"],
    "Education_Level": ["Bachelor", "Master", "PhD", "Bachelor", "Intermediate", "Master", "PhD", "Bachelor"]
})

print(data_ordinal)
```

## 🎯 Convert to Ordered Categories

```python
from pandas.api.types import CategoricalDtype

satisfaction_order = CategoricalDtype(
    categories=["Poor", "Average", "Good", "Excellent"],
    ordered=True
)
data_ordinal["Satisfaction"] = data_ordinal["Satisfaction"].astype(satisfaction_order)
data_ordinal
```

## Sort by satisfaction:

```python
data_ordinal.sort_values("Satisfaction")
```

## Count Satisfaction Values

```python
data_ordinal["Satisfaction"].value_counts()
```

## Compare values:

```python
data_ordinal["Satisfaction"].max()
```

# Complete Data Type Map

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
'name' : ['Ali', 'Sana', 'Bilal'], # Nominal categorical
'grade' : ['A', 'B', 'A'], # Ordinal categorical
'score' : [85, 72, 90], # Discrete numerical
'height_cm' : [170.5, 162.3, 175.8], # Continuous numerical
'temp_c' : [36.6, 37.1, 36.9], # Interval
'income_pkr' : [50000, 80000, 120000], # Ratio
})
```

```python
print(df.dtypes) # shows int64, float64, object
print(df.describe()) # stats for numerical columns only
print(df.nunique()) # unique values per column
```
