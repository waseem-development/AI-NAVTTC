# 📊 Measurement of Central Tendency

> **What is Central Tendency?**
> 
> Imagine you have a group of students and you want to describe their marks with **just one number** — a number that best represents the whole group. That single representative number is called the **measure of central tendency**.
>
> Think of it like asking: *"What is the typical score in this class?"*
>
> There are multiple ways to find this "center" — each has its own strengths and is better suited for different situations. We'll cover:
>
> | Measure | Best For |
> |---|---|
> | **Arithmetic Mean** | Normal data without extreme outliers |
> | **Geometric Mean** | Rates, growth, percentages |
> | **Harmonic Mean** | Speed, ratios, rates |
> | **Median** | Data with outliers |
> | **Mode** | Most frequent value, categories |

```python
import pandas as pd
from scipy import stats as st

# Our dataset: marks of 6 students
# Notice: 100 is an OUTLIER — one student scored way higher than the rest
data = {
    "marks": [10, 20, 20, 30, 40, 100]
}

df = pd.DataFrame(data)
print("Our dataset:")
print(df["marks"].tolist())
print(f"\nTotal students: {len(df)}")
```

---
## 1️⃣ Arithmetic Mean (The Regular Average)

### 📖 Definition
Add up all the values, then divide by the total count.

$$\text{Mean} = \frac{\text{Sum of all values}}{\text{Number of values}} = \frac{10 + 20 + 20 + 30 + 40 + 100}{6}$$

### 🍕 Analogy
Imagine 6 friends order pizza and pay different amounts: ₹10, ₹20, ₹20, ₹30, ₹40, ₹100. The arithmetic mean tells you: *"If everyone had paid equally, how much would each person pay?"* → ₹36.67 each.

### ✅ When to Use
- Calculating **class average marks**
- Finding **average temperature** of a city over a week
- **Average salary** of employees

### ⚠️ Weakness
It gets **pulled by outliers**. Notice how the one student who scored 100 drags the mean up to ~36.67, even though most students scored between 10–40. This is misleading!

```python
arithmetic_mean = df["marks"].mean()

print(f"Arithmetic Mean: {arithmetic_mean:.2f}")
print()
print("Manual check:")
print(f"  Sum = {df['marks'].sum()}")
print(f"  Count = {len(df)}")
print(f"  Mean = {df['marks'].sum()} / {len(df)} = {arithmetic_mean:.2f}")
print()
print(f"⚠️  Most students scored 10-40, but the mean is {arithmetic_mean:.1f} — pulled up by the outlier (100)!")
```

---
## 2️⃣ Geometric Mean

### 📖 Definition
Multiply all values together, then take the **nth root** (where n = number of values).

$$\text{Geometric Mean} = \sqrt[n]{x_1 \times x_2 \times x_3 \times \cdots \times x_n}$$

For our data: $\sqrt[6]{10 \times 20 \times 20 \times 30 \times 40 \times 100}$

### 📈 Analogy
Suppose your money grows every year by different rates:
- Year 1: +10%, Year 2: +20%, Year 3: +50%

If you used arithmetic mean, you'd say "average growth = 26.67%" — but that's **wrong** for compounding! The geometric mean gives the correct average growth rate that, if applied consistently, gives the same final result.

### ✅ When to Use
- **Investment returns** / compound interest (finance)
- **Population growth rates**
- **Averaging percentages and ratios**
- Any situation where values **multiply** together rather than add

### ⚠️ Weakness
Cannot be used if any value is **zero or negative**.

```python
geometric_mean = st.gmean(df["marks"])

print(f"Geometric Mean: {geometric_mean:.2f}")
print()
print("Compare with Arithmetic Mean:")
print(f"  Arithmetic Mean  = {df['marks'].mean():.2f}  ← higher (pulled by outlier)")
print(f"  Geometric Mean   = {geometric_mean:.2f}  ← lower, less affected by extreme values")
print()
print("💡 Geometric Mean is always ≤ Arithmetic Mean")
```

---
## 3️⃣ Harmonic Mean

### 📖 Definition
The reciprocal of the average of reciprocals.

$$\text{Harmonic Mean} = \frac{n}{\frac{1}{x_1} + \frac{1}{x_2} + \cdots + \frac{1}{x_n}}$$

### 🚗 Analogy (The Classic One!)
You drive from City A to City B at **60 km/h**, and return at **40 km/h**.

What's your **average speed** for the whole trip?

- ❌ Wrong (Arithmetic): (60 + 40) / 2 = **50 km/h**
- ✅ Correct (Harmonic): 2 / (1/60 + 1/40) = **48 km/h**

Why? Because you spend *more time* driving slowly, so the slow speed has more impact on the average. Harmonic mean accounts for this automatically.

### ✅ When to Use
- **Average speed** over equal distances
- **Price-to-earnings ratios** in finance
- **F1 Score** in machine learning (balancing precision and recall)
- Any situation involving **rates or ratios**

### ⚠️ Weakness
Cannot be used with **zero values**. Also gives very low weight to large values.

```python
harmonic_mean = st.hmean(df["marks"])

print(f"Harmonic Mean: {harmonic_mean:.2f}")
print()
print("The three means compared:")
print(f"  Arithmetic Mean  = {df['marks'].mean():.2f}  ← largest (most affected by outlier)")
print(f"  Geometric Mean   = {st.gmean(df['marks']):.2f}  ← middle")
print(f"  Harmonic Mean    = {harmonic_mean:.2f}  ← smallest (least affected by outlier)")
print()
print("📌 Rule: Harmonic Mean ≤ Geometric Mean ≤ Arithmetic Mean (always!)")
print("   The more extreme the outlier, the bigger the gap between them.")
```

---
## 4️⃣ Median (The Middle Value)

### 📖 Definition
Sort all values, then pick the **middle one**. If there's an even count, average the two middle values.

Our data sorted: **10, 20, 20, ~~30~~, 40, 100**  
Middle two values (index 3 and 4): **20 and 30** → Median = **(20 + 30) / 2 = 25**

### 🏘️ Analogy
Think about **house prices in a neighborhood**. If most houses cost ₹50 lakh but one billionaire builds a ₹50 crore mansion, the arithmetic mean skyrockets and gives a false picture. The **median** house price still tells you what a *typical* house costs — it ignores that one crazy expensive house.

This is why news reports say *"median household income"* and not *"average household income"* — billionaires would distort the average!

### ✅ When to Use
- **Income / salary data** (where a few very rich people exist)
- **House prices**
- Any dataset with **outliers** or a **skewed distribution**
- When you want the "typical" value

### ✅ Strength
**Not affected by outliers at all.** The 100 in our dataset doesn't move the median.

```python
median = df["marks"].median()

sorted_marks = sorted(df["marks"].tolist())
print(f"Sorted data: {sorted_marks}")
print(f"              {'    '.join(['^' if i in [2,3] else ' ' for i in range(len(sorted_marks))])}")
print(f"              Middle two values are index 3 and 4: 20 and 30")
print()
print(f"Median: {median}")
print()
print(f"🔥 Even though there's a 100 in the data, the median is only {median}!")
print(f"   Compare: Arithmetic Mean = {df['marks'].mean():.2f} (inflated by 100)")
print(f"            Median          = {median} (not affected by 100)")
```

---
## 5️⃣ Mode (The Most Popular Value)

### 📖 Definition
The value that **appears most frequently** in the dataset.

In our data: 10 appears once, **20 appears twice**, 30 appears once, 40 appears once, 100 appears once.

→ Mode = **20** (it appears the most)

### 👟 Analogy
A shoe store wants to know which size to stock the most. They don't care about the average shoe size (7.3 doesn't exist!) or the middle size — they want to know the **most commonly sold size**. That's the mode!

### ✅ When to Use
- **Most popular product** in a store
- **Most common shirt size** to manufacture
- **Favorite color survey** — only mode makes sense here (you can't average colors!)
- **Categorical data** (data that has labels, not numbers)

### 💡 Interesting Facts
- A dataset can have **no mode** (all values unique)
- A dataset can have **multiple modes** (bimodal, trimodal, etc.)
- Mode is the **only measure** that works for non-numeric data

```python
mode = df["marks"].mode()[0]

print(f"Value counts (how many times each mark appears):")
print(df["marks"].value_counts().sort_index())
print()
print(f"Mode: {mode}  ← appears most frequently (2 times)")
```

---
## 📋 Final Summary: All Measures Side by Side

```python
summary = pd.DataFrame({
    "Measure": ["Arithmetic Mean", "Geometric Mean", "Harmonic Mean", "Median", "Mode"],
    "Value": [
        round(df["marks"].mean(), 2),
        round(st.gmean(df["marks"]), 2),
        round(st.hmean(df["marks"]), 2),
        df["marks"].median(),
        df["marks"].mode()[0]
    ],
    "Best For": [
        "General average, symmetric data",
        "Growth rates, ratios, percentages",
        "Speeds, rates, ratios",
        "Skewed data, data with outliers",
        "Most frequent value, categories"
    ],
    "Handles Outliers?": ["❌ No", "⚠️ Somewhat", "✅ Better", "✅ Yes", "✅ Yes"]
})

print("Dataset: [10, 20, 20, 30, 40, 100] — note the outlier: 100")
print("=" * 80)
print(summary.to_string(index=False))
print()
print("💡 Key Takeaway:")
print("   If your data has outliers → use Median")
print("   If your data involves growth/ratios → use Geometric Mean")
print("   If your data involves speeds/rates → use Harmonic Mean")
print("   If you want the most popular value → use Mode")
print("   Otherwise → Arithmetic Mean works fine")
```
