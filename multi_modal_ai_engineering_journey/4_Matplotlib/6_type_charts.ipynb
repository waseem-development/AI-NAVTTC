#### `Different Types of Charts`

### 1- Bar Charts:

```python
plt.bar(
    x,                      # Category Name
    height,                 # Bar Height / Value
    color="Color Name",
    width=value,
    label="Label Name"
)

```
| Parameter | Meaning                    |
| --------- | -------------------------- |
| `x`       | Category names / positions |
| `height`  | Values represented by bars |
| `color`   | Bar color                  |
| `width`   | Width of each bar          |
| `label`   | Name used in the legend    |


````python
import matplotlib.pyplot as plt
import numpy as np
````

````python
products = ["Laptop", "Phone", "Tablet", "Headphones", "Keyboard"]
sales_2026 = [120, 200, 150, 180, 100]
sales_2025 = [100, 180, 130, 160, 90]

plt.bar(products, sales_2026, color="#EF0054", label="Sales 2026")

plt.xlabel("Product", color="#333333")
plt.ylabel("Sales", color="#333333")
plt.title("Product Sale Comparison", fontweight="bold", pad=15, color="#3A3C3A")

plt.legend()

plt.tight_layout()
plt.show()
````

### 2- Pie Charts

``` python
plt.pie(
    values,
    labels=labels_list,
    colors=color_list,
    autopct="1.1f%%"
)
Parameter meaning
```
| Parameter | Meaning                                    |
| --------- | ------------------------------------------ |
| `values`  | Numerical values represented by each slice |
| `labels`  | Names/categories of the slices             |
| `colors`  | Colors of the slices                       |
| `autopct` | Displays percentage on each slice          |


````python
regions = ["North", "South", "East", "West"]
revenue = [3000, 2000, 1500, 1000]
slice_color = ["teal", "skyblue", "lightgreen", "coral"]
plt.pie(
    revenue, 
    labels=regions, 
    autopct="%1.1f%%", 
    colors=slice_color
)
plt.show()
````

### 3- Histogram

``` python

plt.hist(
    data,
    bins = number_of_bins, 
    color = "color_name",
    edgecolor = "black"
    ) 

Parameter meaning

```
| Parameter   | Meaning                                    |
| ----------- | ------------------------------------------ |
| `data`      | Numerical data to be distributed into bins |
| `bins`      | Number of intervals/groups                 |
| `color`     | Color of the histogram bars                |
| `edgecolor` | Color of the edges/borders of each bar     |


````python
scores = [45, 52, 58, 61, 63, 67, 69, 70, 72, 74,
          75, 77, 79, 81, 83, 85, 87, 89, 92, 95]

plt.hist(scores, bins=5, color="orange", edgecolor="black")
plt.xlabel("Scores")
plt.ylabel("Number of Students")
plt.title("Score Distribution of Students")
plt.show()
````

### 4- Scatter Plot

``` python
plt.scatter(
    x,
    y,
    color="color_name",
    marker="style",
    label="label name"
)

Parameter meaning

```
| Parameter | Meaning                   |
| --------- | ------------------------- |
| `x`       | Data for the X-axis       |
| `y`       | Data for the Y-axis       |
| `color`   | Color of the points       |
| `marker`  | Shape/style of the points |
| `label`   | Name used in the legend   |


````python
hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
exam_scores = [45, 52, 58, 65, 70, 76, 84, 91]

social_media_hours = [6, 5.5, 5, 4.5, 4, 3.5, 3, 2.5]
social_media_scores = [45, 48, 52, 58, 62, 68, 75, 82]

plt.scatter(
    hours_studied,
    exam_scores,
    color="green",
    marker="^",
    label="Hours Studied vs Exam Scores"
)

plt.scatter(
    social_media_hours,
    social_media_scores,
    color="red",
    marker="x",
    label="Social Media Use vs Exam Scores"
)

plt.xlabel("Hours")
plt.ylabel("Exam Scores")
plt.legend(loc="best")
plt.title("Study & Social Media Use vs Exam Scores")
plt.grid()

plt.show()
````
