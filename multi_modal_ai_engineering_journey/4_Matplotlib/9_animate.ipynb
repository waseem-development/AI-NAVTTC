```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# data
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots()
line, = ax.plot([], [], lw=2)          # start with an empty line

ax.set_xlim(x.min(), x.max())          # fix axes so they don't jump
ax.set_ylim(y.min() - 0.1, y.max() + 0.1)
ax.grid()

def update(frame):
    line.set_data(x[:frame], y[:frame])   # update BOTH x and y
    return line,

anim = FuncAnimation(
    fig,
    update,
    frames=len(x) + 1,
    interval=30,      # ms between frames
    blit=True
)

plt.close(fig)                          # stops a duplicate static plot
HTML(anim.to_jshtml())                  # shows the animation in the notebook
```
