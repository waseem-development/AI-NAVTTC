### Intro to Subplots

##### `Syntax`
``` python
plt.subplot(
    nrows,
    ncolumns,
    index # Here indexing is not 0-based. It is 1-based
)


````python
import matplotlib.pyplot as plt
import numpy as np
````

````python
languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

width = 0.4

x1 = np.arange(len(languages))

plt.subplot(1, 2, 1)

plt.plot(
    x1,
    repositories,
    color="red",
    linewidth=3,
    linestyle="-.",
    marker="o",
    label="Repositories"
)

plt.xticks(x1, languages, rotation=45, ha="right")

plt.legend()
plt.title("Programming Languages")


plt.subplot(1, 2, 2)

plt.bar(
    x1 + width/2,
    repositories,
    width=width,
    color="blue",
    label="Repositories"
)

plt.xticks(x1, languages, rotation=45, ha="right")

plt.legend()
plt.title("Repositories by Language")

# plt.tight_layout()
plt.show()
````

``` python
fig, ax = plt.subplots(
    nrows,
    ncolumns,
    figsize=(width,height) # Tuple: But this is optional
)

````python
fig, ax = plt.subplots(
    1, 
    2, 
    figsize=(10, 5)
)

ax[0].plot(
        x1, 
        repositories,
        color="green",
        linewidth=3,
        linestyle=":",
        marker="o",
        label="Repositories"
)
ax[0].set_title("Line Graph")

ax[1].bar(
        x1 + width/2, 
        repositories,
        color="teal",
        width=width,
        label="Repositories"
)
ax[1].set_title("Bar Chart")

plt.tight_layout()
plt.show()
````
