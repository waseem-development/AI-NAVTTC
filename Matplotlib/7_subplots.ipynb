### Intro to Subplots

##### `Syntax`
``` python
plt.subplot(
    nrows,
    ncolumns,
    index # Here indexing is not 0-based. It is 1-based
)


````python
import matplotlib.pyplot as plt
import numpy as np
````

````python
languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

plt.subplot(1, 2, 1)

plt.plot(
    languages,
    repositories,
    color="red",
    marker="o",
    label="Repositories"
)

plt.xticks(rotation=45, ha="right")
plt.legend()
plt.title("Programming Languages")


plt.subplot(1, 2, 2)

plt.bar(
    languages,
    repositories,
    color="blue",
    label="Repositories"
)

plt.xticks(rotation=45, ha="right")
# ha="left" → text starts from the tick
# ha="center" → text is centered on the tick
# ha="right" → text ends at the tick
plt.legend()
plt.title("Repositories by Language")

plt.tight_layout()
plt.show()
````

``` python
fig, ax = plt.subplots(
    nrows,
    ncolumns,
    figsize=(width,height) # Tuple: But this is optional
)

````python
languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].plot(
    languages,
    repositories,
    color="green",
    marker="o",
    label="Repositories"
)

ax[0].set_title("Line Graph")
ax[0].set_xticklabels(languages, rotation=45, ha="right")
ax[0].legend()

ax[1].bar(
    languages,
    repositories,
    color="teal",
    label="Repositories"
)

ax[1].set_title("Bar Chart")
ax[1].set_xticklabels(languages, rotation=45, ha="right")
ax[1].legend()

plt.tight_layout()
plt.show()
````
