```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

visitors = [120, 150, 180, 160, 220, 300, 280]
signups = [20, 25, 30, 28, 40, 65, 55]

plt.plot(days, visitors, color="teal", linestyle=":", marker="o", linewidth=3, label="Visitors")
plt.plot(days, signups, color="red", linestyle="--", marker="x", linewidth=3, label="Signup")

plt.title("Daily Visitors and Sign Ups Report")
plt.xlabel("Days")
plt.ylabel("Visitors and Signups")

plt.xticks(
    ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    ["D1", "D2", "D3", "D4", "D5", "D6", "D7"]
    )
plt.yticks([0, 50, 100, 150, 200, 250, 300])

plt.grid(
    axis="y",
    linestyle="--",
    linewidth=0.8,
    alpha=0.5,
    )

plt.ylim(0, 350)

plt.legend()
plt.show()
```
