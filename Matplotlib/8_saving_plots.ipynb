````python
import matplotlib.pyplot as plt
import numpy as np 
````

````python
languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

width = 0.4

x1 = np.arange(len(languages))
````

````python
plt.plot(
    x1,
    repositories,
    color="red",
    linewidth=3,
    linestyle=":",
    marker="o",
    label="Repositories"
)

plt.title("Programming Languages")
plt.legend()

plt.xlabel("Languages")
plt.ylabel("Repositories")

plt.xticks(x1, languages, rotation=45, ha="right")
plt.yticks([0, 500, 1000, 1500], ['Zero', 'Low', 'Medium', 'High'])

plt.tight_layout()
plt.savefig(
    "./outputs/line_graph_high_quality.png",
    dpi=5000,
    bbox_inches="tight"
)
plt.show()
````
