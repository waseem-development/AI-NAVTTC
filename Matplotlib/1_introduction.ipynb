# Import Library

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [16, 40, 26, 50, 45]

plt.plot(x, y)

plt.show()
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

sales = [120, 150, 100, 180, 160]
expenses = [80, 100, 90, 120, 110]

plt.plot(days, sales, label="Sales")
plt.plot(days, expenses, label="Expenses")
plt.title("Sales vs Expenses Stats", fontweight="bold", fontsize=14)

plt.xlabel("Days")
plt.ylabel("Sales and Expenses")

plt.legend()
plt.show()





# plt.plot(days, sales, label="Sales")
# plt.plot(days, expenses, label="Expenses")

# plt.title("Weekly Sales vs Expenses")
# plt.xlabel("Weekdays")
# plt.ylabel("Amount")

# plt.legend()

# plt.show()
```
