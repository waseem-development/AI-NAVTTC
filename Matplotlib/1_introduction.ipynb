```python
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

sales = [120, 150, 100, 180, 160]
expenses = [80, 100, 90, 120, 110]

plt.plot(days, sales, label="Sales")
plt.plot(days, expenses, label="Expenses")

plt.title("Weekly Sales vs Expenses")
plt.xlabel("Weekdays")
plt.ylabel("Amount")

plt.legend()

plt.show()
```
