## Pyplot Functions

````python
import matplotlib.pyplot as plt
````

### `plt.plot()` — Line Plot

```python
plt.plot(
    x,
    y,
    color="color_name",
    linestyle="line_style",
    linewidth=value,
    marker="marker_style",
    label="label_name"
)

````python
months = [

    "Jan", "Feb", "Mar", "Apr", "May", "Jun",

    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"

]

sales = [120, 150, 130, 180, 200, 170, 220, 240, 210, 260, 280, 300]

expenses = [80, 90, 100, 110, 130, 120, 140, 150, 135, 160, 170, 180]

plt.plot(
    months,
    sales,
    color="#fa1e5a",
    marker="o",
    linewidth=3,
    linestyle=":",
    label="Sales"
)
plt.plot(
    months,
    expenses,
    color="#4f7fed",
    marker="o",
    linewidth=3,
    linestyle=":",
    label="Sales"
)
plt.legend(loc="best")

plt.xticks(
    [0, 3, 6, 9],
    ["Q1", "Q2", "Q3", "Q4"]
)

# plt.yticks(
#     [50, 150, 250, 350],
#     ["50-150", "150-250", "250-350", "350+"]
# )

# plt.xlim(0,3)
plt.ylim(0, 400)
plt.grid(
    True,
    axis="x",
    linestyle="--",
    linewidth=4,
    alpha=1
    )
# plt.grid(axis="both") == plt.grid()
plt.show()








# plt.plot(months, sales, color="red", linestyle=":", linewidth=2,
#          marker="o", label="Sales")

# plt.plot(months, expenses, color="blue", linestyle=":", linewidth=2,
#          marker="o", label="Expenses")

# plt.title("Monthly Sales and Expenses Report")

# plt.xlabel("Quarters")

# plt.ylabel("Expenses and Sales")

# plt.legend(loc="lower right")

# plt.xticks(
#     [0, 3, 6, 9],
#     ["Q1", "Q2", "Q3", "Q4"]
# )

# plt.yticks(
#     [50, 150, 250, 350],
#     ["50-150", "150-250", "250-350", "350+"]
# )

# plt.ylim(0, 350)


# # ==================== GRID ====================

# 1. Turn grid ON
# plt.grid()

# 2. Explicitly turn grid ON
# plt.grid(True)

# 3. Turn grid OFF
# plt.grid(False)

# 4. Horizontal grid lines only
# plt.grid(axis="y")

# 5. Vertical grid lines only
# plt.grid(axis="x")

# 6. Both horizontal and vertical grid lines
# plt.grid(axis="both")

# 7. Change grid line style
# plt.grid(linestyle=":")

# 8. Change grid line width
# plt.grid(linewidth=0.5)

# 9. Change grid transparency
# plt.grid(alpha=0.5)

# 10. Combine multiple properties
# plt.grid(
#     axis="both",
#     linestyle=":",
#     linewidth=4,
#     alpha=0.7
# )


# plt.show()
````
