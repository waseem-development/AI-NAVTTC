```python
import numpy as np
import pandas as pd
```

```python
data = {
    "A": [1,2, np.nan, 4, 5],
    "B": [1,np.nan, 3, 4, 5],
    "C": [1,2, 3, np.nan, 5],
    "D": [np.nan,2, 3, 4, 5],
    "E": [1,2, 3, 4, np.nan]
}
df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
df.dropna()
```

```python
df.dropna(thresh=3)  # it says atleast three "nan" values
```

```python
mean = np.mean(df)
df.fillna(mean, inplace=True)
```

```python
mean = np.mean(df)
df.fillna(mean)
```

```python
values = {
    "A": 0,
    "B": 1,
    "C": 2,
    "D": 3,
    "E": 4
}
df.fillna(value=values)
```
