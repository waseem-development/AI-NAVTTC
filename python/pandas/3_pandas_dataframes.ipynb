# 📋 Pandas DataFrames

> **What is a DataFrame?**
>
> A DataFrame is a **2D table** — rows and columns — exactly like an Excel spreadsheet.
>
> ```
>     calories  duration
>  0     420        50
>  1     380        40
>  2     390        45
>  ↑
>  index (row labels)
> ```
>
> Every column in a DataFrame is a **Series**. The DataFrame is just multiple Series glued together.
>
> ### The two main ways to select data:
> | Method | What it does | Example |
> |---|---|---|
> | `df.loc[]` | Select by **label/name** | `df.loc[0]`, `df.loc["Day 1"]` |
> | `df.iloc[]` | Select by **position number** | `df.iloc[0]` = always first row |

````python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data)

print("Our DataFrame:")
print(df)
print()
print(f"Shape: {df.shape}  →  {df.shape[0]} rows, {df.shape[1]} columns")
````

---
## 📍 Selecting Rows with `.loc[]`

`.loc[]` selects rows by their **index label**.

By default the labels are 0, 1, 2... but later we'll give them custom names like "Day 1", "Day 2".

### 🏠 Analogy
`.loc[]` is like finding a person by their **name** in a contact list. You say "give me the row labeled **0**".

````python
# Get a single row by its label
print("Row at label 0:")
print(df.loc[0])
print()
print("Row at label 1:")
print(df.loc[1])
````

````python
# Get MULTIPLE rows — pass a range
# IMPORTANT: df.loc[0:1] includes BOTH 0 and 1 (unlike normal Python slicing!)
print("Rows 0 and 1 (both included):")
print(df.loc[0:1])
print()
print("⚠️  Note: df.loc[0:1] gives 2 rows (0 AND 1). In regular Python, 0:1 would only give index 0!")
````

### 🎯 Getting a Specific Cell

You can pinpoint an exact cell using `.loc[row, column]`.

Like GPS coordinates — row first, then column.

````python
# df.loc[row_label, column_name]
print(f"Row 0, calories column: {df.loc[0, 'calories']}")
print(f"Row 1, duration column: {df.loc[1, 'duration']}")
print()
print("Full table for reference:")
print(df)
````

---
## 🔎 Filtering Rows by Condition

This is one of the most useful features in Pandas — filter rows based on a condition, like asking: **"show me only rows where calories == 420"**.

### 🏠 Analogy
Like using a filter in Excel — you click the dropdown and say "show me only rows where calories > 385". Pandas does this with one line of code.

````python
# Filter: only rows where calories == 420
print("Rows where calories == 420:")
print(df.loc[df['calories'] == 420])
print()

# Filter: only rows where calories > 385
print("Rows where calories > 385:")
print(df.loc[df['calories'] > 385])
print()
print("💡 How this works internally:")
print("  df['calories'] > 385  produces:", list(df['calories'] > 385))
print("  → True/False for each row. .loc[] keeps only the True rows.")
````

---
## 🔢 Selecting by Position with `.iloc[]`

`.iloc[]` selects by **position** — always 0, 1, 2... regardless of what the index labels are.

### `.loc[]` vs `.iloc[]` — What's the difference?

```python
# If your DataFrame has custom labels like "Day 1", "Day 2":
df.loc["Day 1"]   ✅  finds the row LABELED "Day 1"
df.iloc[0]        ✅  finds the FIRST row (position 0), whatever it's labeled
```

**Use `.iloc[]` when you want to say "give me the 3rd row" regardless of its name.**

````python
# iloc[0] always = first row (position 0)
print("First row by position (iloc[0]):")
print(df.iloc[0])
print()

# Pick specific rows by position — pass a list
print("Rows at positions [1, 2, 0] — notice custom order:")
print(df.loc[[1, 2, 0]])  # You can even reorder!
print()
print("💡 df.loc[[1,2,0]] lets you pick rows in any order you want")
````

---
## 🏷️ Named Indexes (Custom Row Labels)

Instead of 0, 1, 2 as row labels, you can give meaningful names like "Day 1", "Day 2", "Day 3".

This makes `.loc[]` much more readable — you access rows by their **meaning**, not just a number.

````python
data1 = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

# Give each row a meaningful name
newDf = pd.DataFrame(data1, index=["Day 1", "Day 2", "Day 3"])

print("DataFrame with named row indexes:")
print(newDf)
print()

# Now access rows by their name!
print('newDf.loc["Day 2"]:')  
print(newDf.loc["Day 2"])
````
