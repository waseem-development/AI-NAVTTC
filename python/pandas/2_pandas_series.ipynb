# Pandas Series

```python
import pandas as pd

a = [100,2,3]

mySeries = pd.Series(a) 
print(mySeries[0]) # PRITNS THE FIRST ELEMENT
mySeries[0] # np.int64(100) type + value
```

## Key/Value Objects as Series

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
myCalSeries = pd.Series(calories)
myCalSeries
```

### Create a Series using only data from "day1" and "day2":

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
myVar = pd.Series(calories, index = ["day1", "day2"])
myVar
```

## Create a DataFrame from two Series:

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
myMergedDataFrame = pd.DataFrame(data)
myMergedDataFrame
```
