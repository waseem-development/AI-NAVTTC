# 📊 Pandas Series

> **What is a Series?**
>
> A Series is like a **single column** from an Excel spreadsheet — a list of values where each value has a label (called an **index**).
>
> ```
> Index | Value
> ------+-------
>   0   |  100     ← index 0
>   1   |    2     ← index 1
>   2   |    3     ← index 2
> ```
>
> By default, Pandas gives numeric indexes starting from 0 — but you can **name them anything** you want.
>
> ### Series vs. Python List
> | Feature | Python List | Pandas Series |
> |---|---|---|
> | Indexes | Only numbers (0, 1, 2...) | Numbers OR custom labels |
> | Math operations | Manual loops needed | Built-in (sum, mean, etc.) |
> | Part of DataFrame? | ❌ No | ✅ Yes — each column is a Series |

There are multiple Ways to Create Series in Pandas

````python
import numpy as np 
import pandas as pd
````

Create Labels

````python
labels = ['a', 'b', 'c']
my_python_list = [10, 20, 30]
my_numpy_array = np.array([10, 20, 30])
d = {"Waseem": 10, "Kamran": 20, "Talha": 30}
````

# Make Series

Series without lables (default series)

````python
pd.Series(my_python_list)
````

Series with custom labels

````python
# pd.Series(my_python_list, labels) possible but not recommended, u should call index manually
pd.Series(my_python_list, index=labels)
````

````python
pd.Series(my_numpy_array)
````

````python
pd.Series(d)
````

````python
a = [100, 2, 3]

mySeries = pd.Series(a)

# Access by index — same as a list
print(f"First element (index 0): {mySeries[0]}")
print()
print("Full Series:")
print(mySeries)
print()
print(f"Data type stored: {mySeries.dtype}   ← np.int64 means 64-bit integer (a number type)")
````

## 🏷️ Custom Indexing in Series

Instead of boring numbers (0, 1, 2), you can give each row a **meaningful label**.

### 🏠 Analogy
Imagine a locker system. Instead of locker numbers 0, 1, 2 — you name them "x", "y", "z". Now you can say "give me what's in locker **x**" instead of "give me locker **0**".

````python
a = [100, 2, 3]

# Pass custom index labels using the index= parameter
mySeries = pd.Series(a, index=["x", "y", "z"])

print("Series with custom labels:")
print(mySeries)
print()
print("Now access by label name instead of number:")
print(f'  mySeries["x"] = {mySeries["x"]}')
print(f'  mySeries["y"] = {mySeries["y"]}')
print(f'  mySeries["z"] = {mySeries["z"]}')
````

## 🗂️ Creating a Series from a Dictionary

A Python **dictionary** (key-value pairs) maps perfectly onto a Series:
- Dictionary **keys** → become the **index**
- Dictionary **values** → become the **data**

### 🍽️ Example: Calorie Tracker
You have calories burned per day stored in a dictionary. Convert it to a Series to get Pandas' built-in analysis tools.

````python
calories = {"day1": 420, "day2": 380, "day3": 390}

myCalSeries = pd.Series(calories)

print("Calories Series (from dictionary):")
print(myCalSeries)
print()
print(f"Total calories burned: {myCalSeries.sum()}")
print(f"Average per day:       {myCalSeries.mean():.1f}")
print(f"Best day (max):        {myCalSeries.idxmax()} → {myCalSeries.max()} calories")
````

### 🔎 Selecting Only Specific Keys from a Dictionary

When creating a Series from a dictionary, you can pass an `index=` list to **select only the entries you want** — and ignore the rest.

This is useful when you have a big dictionary but only need part of it.

````python
calories = {"day1": 420, "day2": 380, "day3": 390}

# Only include day1 and day2 — day3 is ignored
myVar = pd.Series(calories, index=["day1", "day2"])

print("Series with only day1 and day2 (day3 excluded):")
print(myVar)
print()
print("💡 Tip: If you ask for a key that doesn't exist in the dict,")
print("   Pandas will include it but fill with NaN (Not a Number = missing value)")

# Example of NaN when key doesn't exist:
myVarWithNaN = pd.Series(calories, index=["day1", "day4"])  # day4 doesn't exist!
print()
print("Requesting 'day4' which doesn't exist:")
print(myVarWithNaN)
````

## 🏗️ Building a DataFrame from Multiple Series

A **DataFrame** is just multiple Series placed side-by-side as columns.

```
calories  |  duration
----------+-----------
   420    |    50      ← Row 0
   380    |    40      ← Row 1
   390    |    45      ← Row 2
```

You build a DataFrame by passing a **dictionary** where:
- Keys → **column names**
- Values → **lists of data** (which become Series internally)

### 🏠 Analogy
A DataFrame is like an Excel table. Each column is a Series. The whole table together is a DataFrame.

````python
data = {
    "calories": [420, 380, 390],   # ← this becomes one Series (one column)
    "duration": [50, 40, 45]       # ← this becomes another Series (another column)
}

myMergedDataFrame = pd.DataFrame(data)

print("DataFrame created from two lists:")
print(myMergedDataFrame)
print()
print("Accessing the 'calories' column as a Series:")
print(myMergedDataFrame["calories"])
print()
print(f"Each column IS a Series. Type check: {type(myMergedDataFrame['calories'])}")
````
