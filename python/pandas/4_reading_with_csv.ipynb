# 📂 Reading CSV Files in Pandas

> **What is a CSV file?**
>
> CSV stands for **Comma-Separated Values**. It's the simplest way to store tabular data as plain text:
>
> ```
> PROVINCE,DISTRICT,POPULATION
> PUNJAB,Lahore,11126285
> SINDH,Karachi,14916456
> ```
>
> Every spreadsheet app (Excel, Google Sheets) can export to CSV. It's the universal data format.
>
> ### `pd.read_csv()` — your most-used function
> It reads a `.csv` file and instantly gives you a Pandas DataFrame.

````python
import pandas as pd
````

---
##  Controlling How Many Rows Are Displayed

When you have a big dataset (e.g., 528 rows), Pandas doesn't show ALL rows — it abbreviates with `...` in the middle to keep output readable.

The setting `pd.options.display.max_rows` controls how many rows Jupyter shows **on screen**.

### ⚠️ Important!
Changing `max_rows` only affects the **display** — your actual data is untouched. It's like zooming in on a map — the world doesn't change, just how much you can see at once.

````python
# Check the current setting
current_max = pd.options.display.max_rows
print(f"Current max rows displayed: {current_max}")
print()
print("This means if your DataFrame has more than this many rows,")
print("Pandas will show the first and last few, with '...' in between.")
````

````python
# Increase it to show more rows
# Our population dataset has 528 rows, so set it to 530 to see all of them
pd.options.display.max_rows = 530
print(f"New setting: {pd.options.display.max_rows} rows will now be shown")
print()
print("⚠️  Be careful with very large datasets — showing 10,000+ rows will slow down your notebook!")
print("   It's better to use .head(20) or .tail(20) to peek at data instead.")
````

---
## 📖 Reading the CSV File

Now let's actually load the data. `pd.read_csv()` takes a **file path** and returns a DataFrame.

Common parameters you'll use:
- `pd.read_csv("file.csv")` → basic read
- `pd.read_csv("file.csv", nrows=100)` → only read first 100 rows (useful for huge files)
- `pd.read_csv("file.csv", usecols=["col1", "col2"])` → only load specific columns

````python
df = pd.read_csv("./Datasets/population.csv")

print(f"✅ Loaded successfully!")
print(f"   Rows: {df.shape[0]}")
print(f"   Columns: {df.shape[1]}")
print(f"   Columns list: {df.columns.tolist()}")
print()

# Show all rows (since we increased max_rows above)
df
````

---
## 💡 Practical Tips for Reading CSV Files

Here are common real-world scenarios you'll encounter:

````python
# TIP 1: Quick peek without changing display settings
print("=== TIP 1: Use .head(n) to see first n rows ===")
print(df.head(3))
print()

# TIP 2: See the shape (rows, columns)
print("=== TIP 2: Check shape ===")
print(f"df.shape = {df.shape}  →  {df.shape[0]} rows, {df.shape[1]} columns")
print()

# TIP 3: Check for missing data right after loading
print("=== TIP 3: Check for missing values (NaN) ===")
missing = df.isnull().sum()
has_missing = missing[missing > 0]
if len(has_missing) == 0:
    print("✅ No missing values found!")
else:
    print("⚠️  Columns with missing data:")
    print(has_missing)
print()

# TIP 4: Reset display setting back to default when done
pd.options.display.max_rows = 60  # reset to default
print("=== TIP 4: Reset display setting back to default ===")
print(f"max_rows reset to: {pd.options.display.max_rows}")
````
