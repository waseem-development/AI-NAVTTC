# 🐼 Introduction to Pandas

> **What is Pandas?**
>
> Pandas is Python's most popular library for working with **data**. Think of it like Microsoft Excel — but inside Python, way more powerful, and you control everything with code.
>
> The name comes from **"Panel Data"** — a term from statistics for structured data.
>
> ### Why do we need it?
> Raw Python lists and dictionaries are hard to work with when you have thousands of rows of data. Pandas gives you:
> - 📋 **DataFrames** — like spreadsheet tables
> - 📊 **Series** — like a single column of data
> - 🔍 Tools to filter, sort, clean, and analyze data easily
>
> ### 🏠 Analogy
> Imagine you have a huge Excel file with 500 rows of Pakistan population data. In regular Python, accessing and analyzing it would be a nightmare. With Pandas, you can do it in one line.

## 📌 Your First Series

A **Series** is the simplest Pandas object — it's just a list of values with an **index** (row numbers) attached automatically.

Think of it like a single column in Excel:
```
Index | Value
------+------
  0   |   1
  1   |   2
  2   |   3
  3   |   4
  4   |   5
```

````python
import pandas as pd
import os

a = [1, 2, 3, 4, 5]

myFirstSeries = pd.Series(a)

print(f"First element (index 0): {myFirstSeries[0]}")
print()
print("The full Series:")
print(myFirstSeries)
print()
print("Notice: Pandas automatically added index numbers (0, 1, 2, 3, 4) on the left!")
````

## 📥 Importing a Real Dataset

Now the fun part — loading real data. We're using `kagglehub` to download a **Pakistan Population Dataset** from Kaggle (a popular data science platform).

This dataset has population data for every **tehsil (sub-division)** in Pakistan from the 2017 census.

````python
import kagglehub

# Download the dataset — kagglehub caches it locally so it only downloads once
path = kagglehub.dataset_download("mabdullahsajid/population-of-pakistan-dataset")

print("✅ Dataset downloaded!")
print(f"Saved at: {path}")
````

### 📖 Reading the CSV File

`pd.read_csv()` loads a CSV file into a **DataFrame** — a table with rows and columns, just like Excel.

- `df.head()` → shows the **first 5 rows** (like peeking at the top of the file)
- `df.tail()` → shows the **last 5 rows**
- `pd.concat([...])` → **combines** two DataFrames vertically (stacks them)

````python
df = pd.read_csv("./Datasets/population.csv")

dfHead = df.head()   # First 5 rows
dfTail = df.tail()   # Last 5 rows

# Combine first and last 5 rows to see both ends of the data
dfMerged = pd.concat([dfHead, dfTail])

print(f"Total rows in dataset: {len(df)}")
print(f"Total columns: {len(df.columns)}")
print()
print("First 5 + Last 5 rows:")
dfMerged
````

### 💾 Saving to Excel

`df.to_excel()` exports your DataFrame to a `.xlsx` Excel file — great for sharing data with people who use Excel.

- `index=False` → don't include the row numbers (0, 1, 2...) in the file
- `sheet_name=` → name of the Excel sheet tab

````python
# Create the export folder if it doesn't already exist
if not os.path.exists("./Exported_Data"):
    os.makedirs("./Exported_Data")
    print("📁 Created folder: ./Exported_Data")

# Save DataFrame to Excel
df.to_excel("./Exported_Data/new_data.xlsx", index=False, sheet_name="Population_Data")
print("✅ Saved to Excel!")

# Read it back and verify
df_xl = pd.read_excel("./Exported_Data/new_data.xlsx")
print(f"✅ Read back from Excel. Shape: {df_xl.shape}  → ({df_xl.shape[0]} rows, {df_xl.shape[1]} columns)")
````

## 🔍 Exploring the DataFrame

These are the most commonly used tools to **understand a new dataset** quickly.

````python
# .info() → Quick summary: column names, data types, and how many non-null values
# This tells you: "What columns do I have? Any missing data?"
print("=== df.info() ===")
df.info()
````

````python
# Accessing a single column — two ways to do it:
# Way 1: df.COLUMN_NAME  (dot notation — works when column name has no spaces)
# Way 2: df["COLUMN_NAME"]  (bracket notation — always works)

print("All unique provinces in the dataset:")
print(df.PROVINCE.unique())  # .unique() removes duplicates
print()
print("How many times each province appears:")
print(df.PROVINCE.value_counts())
````

````python
# Selecting MULTIPLE columns at once — pass a LIST of column names
# This gives you a smaller table with only the columns you care about
subset = df[["PROVINCE", "DISTRICT", "ALL SEXES (RURAL)"]]
print(f"Showing 3 out of {len(df.columns)} columns:")
subset.head(10)
````

````python
# Useful properties to inspect your DataFrame:
print("All column names:")
print(df.columns.tolist())
print()

# Shape = (rows, columns)
print(f"Shape: {df.shape}")
print(f"  → {df.shape[0]} rows (tehsils/sub-divisions)")
print(f"  → {df.shape[1]} columns")
print()

# len() also gives number of rows
print(f"Number of rows via len(): {len(df)}")
````

````python
# .describe() → Statistical summary of ALL numeric columns at once
# Shows: count, mean, std, min, 25th percentile, median (50%), 75th percentile, max
# This is a great first look at the data to spot unusual values
print("Statistical summary (numeric columns only):")
df.describe()
````
