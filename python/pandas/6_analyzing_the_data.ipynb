## 🗂️ `df.info()` — Understanding the Structure

`df.info()` tells you:
- **Column names** and their **data types** (`int64`, `float64`, `object`)
- **Non-null count** — if a column shows fewer entries than total rows, it has missing data
- **Memory usage** — how much RAM this data takes up

### Data Types Explained:
| Type | Meaning | Example |
|---|---|---|
| `int64` | Whole numbers | 1, 42, 1000 |
| `float64` | Decimal numbers | 3.14, 2.22 |
| `object` | Text / mixed | "PUNJAB", "chicken" |
| `bool` | True/False | survived? |
| `datetime64` | Dates & times | 2024-01-15 |

```python
print("=== Recipe Dataset Info ===")
df_recipe.info()
print()
print("Things to notice:")
print("  - Which columns have 'object' dtype? → those are text")
print("  - Do all columns have the same non-null count? → if not, there's missing data")
```

---
## 📊 `df.describe()` — Statistical Summary

`df.describe()` runs 8 statistics on every numeric column at once.

This is your **instant health check** on the data:
- Are the min/max values reasonable or are there weird outliers?
- Is the mean close to the median (50%)? If not, the data is skewed.
- Are there zeros where there shouldn't be?

### 💡 What to look for:
- `min = 0` in a population column → suspicious (might be missing data coded as 0)
- `max` being extremely large → possible outlier
- `std` being very large compared to `mean` → very spread out data

```python
print("=== Recipe Dataset Statistical Summary ===")
df_recipe.describe()
```

```python
print("=== Population Dataset Statistical Summary ===")
print()

pop_stats = df_population.describe()
print(pop_stats)
print()

print("=== 🔍 Key Observations ===")
print(f"Average rural population per tehsil: {df_population['ALL SEXES (RURAL)'].mean():,.0f}")
print(f"Most populated tehsil (rural):       {df_population['ALL SEXES (RURAL)'].max():,}")
print(f"Least populated tehsil (rural):      {df_population['ALL SEXES (RURAL)'].min():,}")
print()
print("⚠️  Some rows have 0 for urban population — those are rural-only tehsils")
```

## 📋 Quick Reference: EDA Checklist

Every time you get a new dataset, run these 5 things first:

```python
def quick_eda(df, name="Dataset"):
    """Run a quick EDA on any DataFrame"""
    print(f"{'='*50}")
    print(f"  EDA REPORT: {name}")
    print(f"{'='*50}")
    print(f"📐 Shape:          {df.shape[0]} rows × {df.shape[1]} columns")
    print(f"📋 Columns:        {df.columns.tolist()}")
    print()
    
    missing = df.isnull().sum()
    missing_cols = missing[missing > 0]
    if len(missing_cols) == 0:
        print("✅ Missing values: None")
    else:
        print(f"⚠️  Missing values in {len(missing_cols)} column(s):")
        print(missing_cols)
    print()
    
    numeric_cols = df.select_dtypes(include='number').columns
    text_cols = df.select_dtypes(include='object').columns
    print(f"🔢 Numeric columns ({len(numeric_cols)}): {list(numeric_cols)}")
    print(f"📝 Text columns   ({len(text_cols)}):  {list(text_cols)}")
    print(f"{'='*50}")

# Run it on both datasets
quick_eda(df_population, "Pakistan Population")
print()
quick_eda(df_recipe, "US Recipes")
```
