```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'B', 'A', 'C', 'A', 'B', 'C', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
```

```python
df
```

##### Group by Category

```python
cat = df.groupby("Category")
cat ## cannot see it because groupby returns an object not a dataframe
```

```python
for i, v in cat:
    print(i)
    print(v)
```

##### Group by Category and calculate the some of Sales

```python
cat1 = df.groupby("Category")["Sales"].sum()
cat1
```

##### Group by store and calculate the some of Sales

```python
cat = df.groupby("Store")["Sales"].sum()
cat
```

## - Group by multiple columns
## - Group by category and store

```python
cat = df.groupby(["Store", "Category"])["Sales"].sum()
cat
```

## Aggregation

### It means to combine different things to get a single result

```python
df["Sales"].mean()
# mean median min max count std
```

```python
df["Sales"].agg(["mean", "median", "min", "max", "std"])
```
