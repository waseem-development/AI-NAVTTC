# NumPy Data Types

NumPy refers to data types with short character codes. Here's every one, explained properly.

---

## `i` — Integer

Whole numbers. No decimal point. Your everyday counting numbers — can be negative.

```python
x = np.array([1, 2, 3], dtype='i')
# Good for: age, pixel coordinates, count of items
```

---

## `b` — Boolean

Only two possible values: `True` or `False`. Internally stored as `1` and `0`.

```python
x = np.array([True, False, True], dtype='b')
```

You use this constantly in CV. When you do `img > 128`, NumPy returns a boolean array — every pixel either passes the condition or doesn't.

---

## `u` — Unsigned Integer

Like `i` but **no negatives**. Range starts from 0.

```python
x = np.array([0, 128, 255], dtype='u1')  # u1 = 1 byte → range 0–255
```

This is exactly what image pixels are. When you check `img.dtype` on your Japan street image, it returns `uint8` — unsigned integer, 8 bits, range 0 to 255. That's why pixel values are always 0–255.

---

## `f` — Float

Numbers with a decimal point.

```python
x = np.array([0.5, 1.7, 3.14], dtype='f')
# Good for: normalized pixel values (0.0–1.0), model weights, loss values
```

In deep learning you often convert images from `uint8` (0–255) to `float32` (0.0–1.0) by dividing by 255:

```python
img_float = img.astype('float32') / 255.0
```

---

## `c` — Complex Float

Numbers with a real and imaginary part, like `3 + 2j`. Rarely used in everyday work.

```python
x = np.array([1+2j, 3+4j], dtype='c')
```

You might see this in CV when doing frequency-domain operations on images (like Fourier transforms).

---

## `m` — Timedelta

Represents a **duration** — a difference between two points in time.

```python
x = np.array([10, 20], dtype='m8[s]')  # durations in seconds → "10 seconds", "20 seconds"
# Good for: measuring time between frames in a video
```

---

## `M` — Datetime

An actual **point in time** — a specific date and/or timestamp.

```python
x = np.array(['2024-01-01', '2024-06-15'], dtype='M')
# Good for: timestamps on sensor data, video recordings
```

> Easy way to remember: `m` (small) = duration. `M` (capital) = actual moment in time.

---

## `O` — Object

A catch-all. When NumPy can't fit data into a clean numeric type, it stores it as a generic Python object.

```python
x = np.array([1, 'hello', [1, 2, 3]], dtype='O')
```

You lose all NumPy speed advantages. If you see `dtype=object` in your array, it's usually a sign something is wrong with your data.

---

## `S` — String (bytes)

Fixed-length strings stored as raw bytes.

```python
x = np.array(['cat', 'dog'], dtype='S')
# Stored as: b'cat', b'dog'  ← notice the b prefix
```

Not used much anymore — `U` has replaced it for most purposes.

---

## `U` — Unicode String

Strings that support all languages and special characters.

```python
x = np.array(['hello', 'جاپان', 'tokyo'], dtype='U')
# Can store Arabic, Urdu, Japanese — any language
```

This is the string type you actually want to use.

---

## `V` — Void

A raw block of memory with no specific type assigned. Just holds bytes without caring what they mean.

```python
x = np.array([(1, 2.5)], dtype='V')
# Rarely used directly — shows up in structured/record arrays
```

---

## The 3 You'll Actually Use in CV

```python
uint8    # 'u1' → pixel values 0–255, what cv2.imread() gives you
float32  # 'f4' → normalized 0.0–1.0, what models expect
bool     # 'b'  → masks and thresholding (e.g. img > 128)
```

````python
import numpy as np
i = np.array([1,2,3], dtype='i')
print(f"i = {i}")
print(f"i.dtype = {i.dtype}")
````

````python
b = np.array([True, True, False], dtype='b')
print(f"b = {b}")
print(f"b.dtype = {b.dtype}")
````

````python
u = np.array([1,2,3], dtype='u1')
print(f"u = {u}")
print(f"u.dtype = {u.dtype}")
````

````python
f = np.array([1.4,2.5,3.14], dtype='f')
print(f"f = {f}")
print(f"f.dtype = {f.dtype}")
````

````python
c1 = np.array([1+2j, 3+4j], dtype='c8')   # complex64  (2× float32)
c2 = np.array([1+2j, 3+4j], dtype='c16')  # complex128 (2× float64)

print(f"c1 = {c1}")
print(f"c1.dtype = {c1.dtype}")

print(f"c2 = {c2}")
print(f"c2.dtype = {c2.dtype}")
````

````python
m = np.array([10, 20], dtype='m8[s]')  # durations in seconds → "10 seconds", "20 seconds"
print(f"m = {m}")
print(f"m.dtype = {m.dtype}")
````

````python
M = np.array(['2026-04-04', '2004-06-06'], dtype='M')
print(f"M = {M}")
print(f"M.dtype = {M.dtype}")
````

````python
O = np.array([1, 'hello', [1, 2, 3]], dtype='O')
print(f"O = {O}")
print(f"O.dtype = {O.dtype}")
````

````python
S = np.array(["Cat", "Dog"])
print(f"S = {S}")
print(f"S.dtype = {S.dtype}")
````

````python
U = np.array(["جاپان", "India", "パキスタン"])
print(f"U = {U}")
print(f"U.dtype = {U.dtype}")
````

````python
import struct

raw = struct.pack('if', 1, 2.5)  # pack int + float into bytes
V = np.frombuffer(raw, dtype='V8')
print(f"V = {V}")
print(f"V.dtype = {V.dtype}")
````
