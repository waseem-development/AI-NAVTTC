# Numpy Basics


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(f"Numpy array (arr): {arr}")
```

## Checking Numpy Version


```python
print(f"Numpy version: {np.__version__}")
```

## Type of Numpy Array

---

- The array object in NumPy is called ndarray.
- We can create a NumPy ndarray object by using the array() function.

---


```python
print(f"Type of Numpy array (ndarray): {type(arr)}")
```

### To create an ndarray, we can pass a list, tuple or any array-like object into the array() method, and it will be converted into an ndarray:


```python
my_tuple = (1, 2, 3, 4)
tup_array = np.array(my_tuple)
print(f"Array created from Tuple: {tup_array}")
```

```python
my_list = [1, 2, 3, 4]
list_array = np.array(my_list)
print(f"Array created from List: {list_array}")
```

```python
strr = "Hey there"
str_array = np.array(strr)
print(f"Array created from String: {str_array}")
```

### String → 0-dimensional array because NumPy treats the entire string as a single scalar object, not individual characters. It doesn't split "Hey there" into ['H', 'e', 'y', ...].

- You can verify this:


```python
string_array = np.array("Hey there")
print(f"Shape of the array: {string_array.shape}")  # ()  ← 0-d array, no dimensions
print(f"Number of dimensions: {string_array.ndim}")  # 0
print(
    f"Data Type of the array: {string_array.dtype}"
)  # <U9  (Unicode string of length 9)
```

```python
string_array_from_list = np.array(list("Hey there"))
print(f"Array (string_array_from_list): {string_array_from_list}")
print(f"Shape of the array: {string_array_from_list.shape}")
print(f"Number of dimensions: {string_array_from_list.ndim}")
print(f"Data Type of the array: {string_array_from_list.dtype}")
```

## Types of Arrays


#### 1. 0D Array: An array with exactly one single value


```python
zero_d_array = np.array(42)
print(f"Array (zero_d_array): {zero_d_array}")
print(f"Shape of the array: {zero_d_array.shape}")
print(f"Dimensions of the array: {zero_d_array.ndim}")
print(f"Data Type of the array: {zero_d_array.dtype}")
print(f"Size(number of elements) of the array: {zero_d_array.size}")
print(f"Memory in bytes used by one element. int32 = 4 bytes, float64 = 8 bytes, etc. for the array: {zero_d_array.itemsize}")
print(f"Total memory = size x itemsize. Useful for checking how much RAM your array uses for the array: {zero_d_array.nbytes}")
```

#### 2. 1D Array (Vector): An array with exactly one single row


```python
one_d_array = np.array([10, 20, 30, 40])
print(f"Array (one_d_array): {one_d_array}")
print(f"Shape of the array: {one_d_array.shape}")
print(f"Dimensions of the array: {one_d_array.ndim}")
print(f"Data Type of the array: {one_d_array.dtype}")
print(f"Size(number of elements) of the array: {one_d_array.size}")
print(f"Memory in bytes used by one element. int32 = 4 bytes, float64 = 8 bytes, etc. for the array: {one_d_array.itemsize}")
print(f"Total memory = size x itemsize. Useful for checking how much RAM your array uses for the array: {one_d_array.nbytes}")
```

#### 3. 2-D Arrays

---

- An array that has 1-D arrays as its elements is called a 2-D array.
- These are often used to represent matrix or 2nd order tensors.

---


```python
two_d_array = np.array(
    [
        [10, 20, 30, 40],
        [100, 200, 300, 400],
        [1000, 2000, 3000, 4000],
        [10000, 20000, 30000, 40000],
        [100000, 200000, 300000, 400000],
    ]
)
print(f"Array (two_d_array): \n{two_d_array}")
print(f"Shape of the array: {two_d_array.shape}")
print(f"Dimensions of the array: {two_d_array.ndim}")
print(f"Data Type of the array: {two_d_array.dtype}")
print(f"Size(number of elements) of the array: {two_d_array.size}")
print(f"Memory in bytes used by one element. int32 = 4 bytes, float64 = 8 bytes, etc. for the array: {two_d_array.itemsize}")
print(f"Total memory = size x itemsize. Useful for checking how much RAM your array uses for the array: {two_d_array.nbytes}")
```

#### 3. 3-D arrays

---

- An array that has 2-D arrays (matrices) as its elements is called 3-D array.
- These are often used to represent a 3rd order tensor.
- (Stack of matrices — like an RGB image (height, width, channels=3). Three axes.)

---


```python
three_d_array = np.array(
    [
        [
            [10, 20, 30, 40],
            [100, 200, 300, 400],
            [1000, 2000, 3000, 4000],
            [10000, 20000, 30000, 40000],
            [100000, 200000, 300000, 400000],
        ],
        [
            [1.1, 2.1, 3.1, 4.1],
            [100.1, 200.1, 300.1, 400.1],
            [1000.1, 2000.1, 3000.1, 4000.1],
            [10000.1, 20000.1, 30000.1, 40000.1],
            [100000.1, 200000.1, 300000.1, 400000.1],
        ],
    ]
)
print(f"Array (three_d_array): \n{three_d_array}")
print(f"Shape of the array: {three_d_array.shape}")
print(f"Dimensions of the array: {three_d_array.ndim}")
print(f"Data Type of the array: {three_d_array.dtype}")
print(f"Size(number of elements) of the array: {three_d_array.size}")
print(f"Memory in bytes used by one element. int32 = 4 bytes, float64 = 8 bytes, etc. for the array: {three_d_array.itemsize}")
print(f"Total memory = size x itemsize. Useful for checking how much RAM your array uses for the array: {three_d_array.nbytes}")
```

#### 4. 5-D Arrays: Used in video data, 3D scans, or deep architectures. Rare in everyday code.

```python
# Video batch: (samples, frames, H, W, C)
vid_batch = np.zeros((4, 30, 224, 224, 3))
print(f"Array (four_d_array): \n{vid_batch}")
print(f"Shape of the array: {vid_batch.shape}") # (4, 30, 224, 224, 3)
# 4 videos, 30 frames each, 224×224 RGB
print(f"Dimensions of the array: {vid_batch.ndim}")  # 5
print(f"Data Type of the array: {vid_batch.dtype}")
print(f"Size(number of elements) of the array: {vid_batch.size}")
print(f"Memory in bytes used by one element. int32 = 4 bytes, float64 = 8 bytes, etc. for the array: {vid_batch.itemsize}")
print(f"Total memory = size x itemsize. Useful for checking how much RAM your array uses for the array: {vid_batch.nbytes}")
```
