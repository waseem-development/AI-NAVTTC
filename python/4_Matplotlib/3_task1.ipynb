*Importing Matplotlib Library*

```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [100, 130, 120, 160, 180, 170, 200, 220, 210, 250, 270, 300]
cost = [60, 70, 75, 90, 100, 95, 110, 120, 115, 130, 145, 150]


plt.plot(months, revenue, color="black", linestyle=":", linewidth=3, label="Revenue")
plt.plot(months, cost, color="red", linestyle=":", linewidth=3, label="Amount")
plt.xlabel("Quarters")
plt.ylabel("Amount")
plt.title("Monthly Revenue & Cost")
plt.legend(loc="upper left")
plt.xticks(
    [0, 3, 6, 9],
    ["Q1", "Q2", "Q3", "Q4",])

plt.yticks(
    [50, 150, 250, 350],
    ["50-150", "150-250", "250-350", "350+"])

plt.grid(axis="y", linestyle="--")
plt.ylim(0,350)
```
