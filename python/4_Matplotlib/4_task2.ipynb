```python
import matplotlib.pyplot as plt
```

```python
subjects = ["Math", "Physics", "Computer Science", "English", "Artificial Intelligence", "Database"]

student_a = [75, 82, 90, 68, 95, 88]
student_b = [80, 78, 85, 75, 90, 92]

plt.plot(subjects, student_a, color="teal", linestyle=":", marker="o", linewidth=3, label="Ahmed\' Marks")
plt.plot(subjects, student_b, color="red", linestyle=":", marker="x", linewidth=3, label="Bilal\' Marks")

plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.xticks(
    ["Math", "Physics", "Computer Science", "English", "Artificial Intelligence", "Database"],
    ["M", "Phy", "CS", "Eng", "AI", "DB"]
)
plt.yticks(
    [0, 20, 40, 60, 80, 100]
)
plt.grid(linewidth=3, linestyle=":", alpha=0.7)
plt.show()
```
