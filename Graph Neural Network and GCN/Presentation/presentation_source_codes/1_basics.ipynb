```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [16, 40, 26, 50, 45]

plt.plot(x, y)

plt.show()
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

sales = [120, 150, 100, 180, 160]
expenses = [80, 100, 90, 120, 110]

plt.plot(days, sales, label="Sales")
plt.plot(days, expenses, label="Expenses")
plt.title("Weekly Sales vs Expenses")
plt.xlabel("Weekdays")
plt.ylabel("Amount")

plt.legend()
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
"Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [120, 150, 130, 180, 200, 170,
220, 240, 210, 260, 280, 300]
expenses = [80, 90, 100, 110, 130, 120,
140, 150, 135, 160, 170, 180]

plt.plot(months, sales, color="red", linestyle=":",
linewidth=2, marker="o", label="Sales")
plt.plot(months, expenses, color="blue", linestyle=":",
linewidth=2, marker="o", label="Expenses")
plt.title("Monthly Sales and Expenses Report")
plt.xlabel("Quarters")
plt.ylabel("Expenses and Sales")

plt.legend(loc="lower right")
plt.xticks(
    [0, 3, 6, 9], 
    ["Q1", "Q2", "Q3", "Q4"])
plt.yticks(
    [50, 150, 250, 350],
["50-150", "150-250", "250-350", "350+"])

plt.ylim(0, 350)
plt.grid(axis="both", linestyle=":", linewidth=4, alpha=0.7)
plt.show()
```
