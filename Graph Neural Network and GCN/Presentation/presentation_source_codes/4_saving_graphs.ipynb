```python
import matplotlib.pyplot as plt

languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

plt.plot(
    languages,
    repositories,
    color="red",
    marker="o",
    label="Repositories"
)

plt.title("Programming Languages")
plt.legend()
plt.xlabel("Languages")
plt.ylabel("Repositories")

plt.xticks(rotation=45, ha="right")

plt.yticks(
    [0, 500, 1000, 1500],
    ["Zero", "Low", "Medium", "High"]
)

plt.tight_layout()

plt.savefig(
    "./outputs/line_graph_high_quality.png",
    dpi=500,
    bbox_inches="tight"
)

plt.show()
```
