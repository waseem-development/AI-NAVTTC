```python
import matplotlib.pyplot as plt
```

```python
languages = ["Python", "Java", "Rust", "JavaScript", "C++", "Go"]
repositories = [65, 120, 30, 95, 45, 80]

plt.subplot(2, 2, 1)

plt.plot(
    languages,
    repositories,
    color="red",
    marker="o",
    label="Repositories"
)

plt.xticks(rotation=45, ha="right")
plt.legend()
plt.title("Programming Languages")


plt.subplot(2, 2, 2)

plt.bar(
    languages,
    repositories,
    color="blue",
    label="Repositories"
)

plt.xticks(rotation=45, ha="right")
plt.legend()
plt.title("Repositories by Language")


plt.subplot(2, 2, 3)

plt.scatter(
    languages,
    repositories,
    color="green",
    label="Repositories"
)

plt.xticks(rotation=45, ha="right")
plt.legend()
plt.title("Repository Distribution")


plt.subplot(2, 2, 4)

plt.pie(
    repositories,
    labels=languages,
    autopct="%1.0f%%",
    colors=["teal", "skyblue", "lightgreen", "coral", "orange", "pink"]
)

plt.title("Repository Share")


plt.tight_layout()
plt.show()
```

```python
languages = ["Go", "Python", "Rust", "JavaScript", "C++", "Java"]
repositories = [350, 850, 280, 720, 480, 610]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(
    languages,
    repositories,
    color="green",
    marker="o",
    linewidth=2,
    label="Repositories"
)
ax[0, 0].set_title("Line Graph")
ax[0, 0].set_xlabel("Languages")
ax[0, 0].set_ylabel("Repositories")
ax[0, 0].tick_params(axis="x", rotation=45)
ax[0, 0].legend()
ax[0, 0].grid(True, linestyle=":", alpha=0.5)

ax[0, 1].bar(
    languages,
    repositories,
    color="teal",
    label="Repositories"
)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_xlabel("Languages")
ax[0, 1].set_ylabel("Repositories")
ax[0, 1].tick_params(axis="x", rotation=45)
ax[0, 1].legend()
ax[0, 1].grid(axis="y", linestyle=":", alpha=0.5)

ax[1, 0].scatter(
    languages,
    repositories,
    color="red",
    marker="o",
    label="Repositories"
)
ax[1, 0].set_title("Scatter Plot")
ax[1, 0].set_xlabel("Languages")
ax[1, 0].set_ylabel("Repositories")
ax[1, 0].tick_params(axis="x", rotation=45)
ax[1, 0].legend()
ax[1, 0].grid(True, linestyle=":", alpha=0.5)

ax[1, 1].pie(
    repositories,
    labels=languages,
    autopct="%1.1f%%"
)
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```
