# Getting Started with Data visualization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Matplotlib basics

```python
x = np.linspace(1, 10, 10)
y = x**2
z = x**3
a = y**2
b = a**2
c = b**2
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y)
plt.title("Hello World in Matplotlib")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

### Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(z,a)
plt.subplot(2,2,3)
plt.plot(b,c)
plt.subplot(2,2,4)
plt.plot(x,c)
```
