# Distribtion Plots

These are used for quantitative variables. i.e. for the visualisation of their distribuion.
It Includes
- displot / histplot
- join plots
- pair plots
- rugplots

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```python
df = sns.load_dataset('tips')
df
```

```python
df.shape
```

```python
df.describe()
```

```python
df.info()
```

```python
df["size"].unique()
```

```python
sns.displot(df["total_bill"], bins=30, kde=True) #kernel density estimator
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
sns.histplot(df["total_bill"], bins=20, kde=True)

plt.subplot(1,2,2)
sns.histplot(df["tip"], bins=20, kde=True)

plt.tight_layout()
plt.show()
```

## Jointplot

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="scatter")
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="kde")
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="hist")
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="hex")
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="reg")
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="resid")
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue="sex")
```

```python
sns.pairplot(df, hue="sex", palette="rainbow")
```

```python
sns.rugplot(df)
```

```python
sns.rugplot(df["total_bill"])
```
