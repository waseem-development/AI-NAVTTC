# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

- boxplot

- violinplot

- stripplot

- swarmplot

- barplot

- countplot

Let's go through examples of each!

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.countplot(x=df["sex"], hue=df["smoker"], palette='rainbow')
```

```python
sns.barplot(x=df["sex"], y=df["total_bill"])
```

```python
sns.barplot(x=df["sex"], y=df["total_bill"], estimator=np.sum)
```

```python
sns.barplot(x=df["sex"], y=df["tip"])
```

```python
sns.boxplot(x='tip', y='day', data=df, palette="Set2")
```

```python
sns.violinplot(x=df["day"], y=df["total_bill"], palette="RdBu")
```

```python
sns.swarmplot(x=df["tip"], y=df["day"], palette="PiYG")
```

```python
sns.stripplot(x=df["tip"], y=df["day"], palette="Spectral")
```

### Combining Violen and Swarm Plots

```python
sns.violinplot(x=df["tip"], y=df["day"], palette="rainbow")
sns.swarmplot(x=df["tip"], y=df["day"], palette="rainbow")
```
