```python
xaxis = np.linspace(0, 5, 11)
```

```python
fig, ax = plt.subplots(figsize=(12,6))

# linewidth = thickness of the line (0.25 is very thin, 2.00 is thick)
ax.plot(xaxis, xaxis*1, color="red", linewidth=0.25)
ax.plot(xaxis, xaxis*2, color="red", linewidth=0.50)
ax.plot(xaxis, xaxis*3, color="red", linewidth=1.00)
ax.plot(xaxis, xaxis*4, color="red", linewidth=2.00)

# linestyle (or ls) = the pattern/style of the line:
#   '-'    → solid line        ──────────
#   '--'   → dashed line       ── ── ── ──
#   '-.'   → dash-dot line     ──·──·──·──
#   ':'    → dotted line       ············
#   'steps'→ staircase line
# lw = shorthand for linewidth
ax.plot(xaxis, xaxis*5, color="green", lw=3, linestyle='-')
ax.plot(xaxis, xaxis*6, color="green", lw=3, ls='-.')
ax.plot(xaxis, xaxis*7, color="green", lw=3, ls=':')

# custom dash: set_dashes([line_length, space_length, line_length, space_length, ...])
# [5, 50, 15, 50] means: draw 5pts, skip 50pts, draw 15pts, skip 50pts, repeat
line, = ax.plot(xaxis, xaxis*8, color="black", lw=1.50)
line.set_dashes([5, 50, 15, 50])

# marker = the symbol drawn AT each data point on the line:
#   '+'  → plus sign      +
#   'o'  → circle         ●
#   '*'  → star           ★
#   's'  → square         ■
#   ','  → pixel (tiny dot)
#   '.'  → small dot      •
#   '1'  → tri-down       ▼ (like a Y shape pointing down)
#   '2'  → tri-up         △
#   '3'  → tri-left       ◁
#   '4'  → tri-right      ▷
#   'D'  → diamond        ◆
#   '^'  → triangle up    △
#   'v'  → triangle down  ▽
#   '<'  → triangle left  ◁
#   '>'  → triangle right ▷
#   'x'  → X mark         ✕
#   'p'  → pentagon       ⬠
#   'h'  → hexagon        ⬡

# lw=3 → thick line, ls='-' → solid, marker='+' → plus sign at each data point
ax.plot(xaxis, xaxis*9,  color="blue",   lw=3, ls='-',  marker='+')
ax.plot(xaxis, xaxis*10, color="blue",   lw=3, ls='--', marker='o')   # circle marker
ax.plot(xaxis, xaxis*11, color="blue",   lw=3, ls='-',  marker='s')   # square marker
ax.plot(xaxis, xaxis*12, color="blue",   lw=3, ls='--', marker='1')   # tri-down marker

# markersize (or ms) = size of the marker symbol in points
# lw=1 → thin line so the markers stand out more
ax.plot(xaxis, xaxis*13, color="purple", lw=1, ls='-',  marker='o', markersize=2)  # tiny circle
ax.plot(xaxis, xaxis*14, color="purple", lw=1, ls='-',  marker='o', markersize=4)  # medium circle
ax.plot(xaxis, xaxis*15, color="purple", lw=1, ls='-',  marker='o', markersize=8,  # large circle
        markerfacecolor="red")          # markerfacecolor = fill color INSIDE the marker

# markeredgewidth = thickness of the border/outline around the marker
# markeredgecolor = color of that border/outline
# markerfacecolor = fill color inside the marker
ax.plot(xaxis, xaxis*16, color="purple", lw=1, ls='-',  marker='s', markersize=8,
        markerfacecolor="yellow",    # square filled with yellow
        markeredgewidth=3,           # thick border around the square
        markeredgecolor="green")     # border is green
```
