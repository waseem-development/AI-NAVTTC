# OpenCV Basics: Remaining topics

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("./Images/fruit.png")

if img is None: 
    print("Image not found")
else: 
    cv2.imshow("Fruit", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
```

#### *Save Images*: 
- Syntax`cv2.imwrite(output.jpg, image)`

```python
if img is not None:
    success = cv2.imwrite("./Outputs/saved_japanese.jpg", img)
    if success:
        print("Saved succesully as \"/Outputs/saved_japanese.jpg\"")
    else:
        print("Failed to save the image")
else:
    print("Error: Could not load image")
```

#### *Image Dimensions*

- Use the *`.shape`* method. (height, width, channel)
> If our image is grayscale then channel parameter will be missing and only width and height will be displayed

```python
if img is not None:
    h, w, c = img.shape
    print(f"Image loaded: \nHeight: {h}\nWidth: {w}\nChannel: {c}")
else: 
    print("Could not load image")
```

#### *Grayscale Conversion:* To convert an image to grayscale
*Why?* Because when we are working on a face-detection or object-tracking project then these work efficiently with grayscale images
`cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)`

```python
if img is not None:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    cv2.imshow("Grayscale Image:", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else: 
    print("Could Not load image")
```
